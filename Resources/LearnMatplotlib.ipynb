```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Type of Plots

## Line plot

```python
x = [1,2,3,4,5]
y = [10,20,25,30,40]
```

```python
plt.plot(x,y);
plt.title("Simple Line Plot")
plt.xlabel("X-axis Label");
plt.ylabel("Y-axis Label");
plt.grid(True)
```

## Bar Chart

```python
months = ['Jan','Feb','Mar','Apr']
revenue = [100,200,250,500]
```

```python
plt.figure(figsize=(10,6))
plt.bar(months,revenue)
plt.title("Monthly Revenue Bar Chart")
plt.xlabel("Months");
plt.ylabel("revenue");
```

## Scatter Plot

```python
heights = [65,70,68,72,66]
weights = [150,160,155,180,145]
```

```python
plt.scatter(heights,weights)
plt.title('Heights vs. Weights Scatter Plot')
plt.xlabel('Height (inches)');
plt.ylabel('weight (pounds)');
```

## Subplots and Legends

```python
days = [1,2,3,4,5]
original_price = [10,20,25,30,40]
predicted_price = [8,21,25,28,35]
```

```python
plt.plot(days, original_price,'b-o', label='Original Price')
plt.plot(days, predicted_price, 'r-o', label='Predicted Price')
plt.legend()
plt.title('Plot with Legends');
plt.xlabel('Weekdays');
plt.ylabel('Price');
```

## Subplots

```python
# First subplot - Original price
plt.subplot(3,1,1)
plt.plot(days, original_price,'b-o', label='Original Price')
plt.legend()
plt.title('Original prices over time');
plt.xlabel('Weekdays');
plt.ylabel('Price');

# Second subplot - Predicted price
plt.subplot(3,1,2)
plt.plot(days, predicted_price, 'r-o', label='Predicted Price')
plt.legend()
plt.title('Predicted prices over time');
plt.xlabel('Weekdays');
plt.ylabel('Price');

# Third subplot - difference between Oirginal price & Predicted price
price_difference = [o - p for o, p in zip(original_price,predicted_price)]
plt.subplot(3,1,3)
plt.bar(days, price_difference, color='yellow', label='Price difference')
plt.legend()
plt.title('Predicted prices over time');
plt.xlabel('Weekdays');
plt.ylabel('Price');
plt.tight_layout()
```

## Saving the figure

```python
plt.plot(days, original_price,'b-o', label='Original Price')
plt.plot(days, predicted_price, 'r-o', label='Predicted Price')
plt.legend()
plt.title('Plot with Legends');
plt.xlabel('Weekdays');
plt.ylabel('Price');
plt.savefig('example_predicted.png',dpi=300) # Save the Image
```
